```python
import stanza

# ---------------------------
# Inicializar el pipeline en español
# ---------------------------
def init_pipeline():
    print("Descargando modelos de Stanza (solo la primera vez)...")
    stanza.download('es')  # Descarga el modelo en español si no está
    print("Inicializando pipeline...")
    return stanza.Pipeline(lang='es', processors='tokenize,mwt,pos,lemma')

# ---------------------------
# Función para lematizar una oración
# ---------------------------
def lemmatize_sentence(pipeline, sentence):
    doc = pipeline(sentence)
    lemmas = []
    for sent in doc.sentences:
        for word in sent.words:
            lemmas.append(word.lemma)
    return lemmas

# ---------------------------
# Main
# ---------------------------
if __name__ == "__main__":
    pipeline = init_pipeline()

    # Frase de ejemplo (puedes cambiarla)
    input_sentence = "Los niños estaban jugando en el parque con sus amigos."

    print(f"\nFrase original: {input_sentence}")

    lemmas = lemmatize_sentence(pipeline, input_sentence)
    print("\nLematización palabra por palabra:")
    print(lemmas)
```
